```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../Data/raw/customer_churn_dataset-1.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
numeric_columns = [
    "Age",
    "Monthly_Charges",
    "Avg_Resolution_Hours",
    "NPS_Score"
]

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()

```

```python
print("Final dataset shape:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
```

```python
df.dtypes
```

```python
print("Subscription Plans:")
print(df["Subscription_Plan"].unique())

print("\nBilling Frequencies:")
print(df["Billing_Frequency"].unique())

print("\nChurn Values:")
print(df["Churn"].unique())
```

```python
df["Churn"] = df["Churn"].map({"No": 0, "Yes": 1})
df["Churn"].value_counts()
```

```python
df.to_csv("../Data/processed/customer_churn_processed.csv", index=False)
```
