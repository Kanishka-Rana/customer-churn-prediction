```python
#Load Data 
import pandas as pd
import numpy as np

df = pd.read_csv("../Data/processed/customer_churn_processed.csv")
```

```python
#7D vs 30D login feature 
df["Expected_Login_7D"] = df["Login_Count_30D"] / 30 * 7

df["Login_Drop_Percent"] = (
    (df["Expected_Login_7D"] - df["Login_Count_7D"])
    / df["Expected_Login_7D"]
) * 100

df["Login_Drop_Percent"] = df["Login_Drop_Percent"].clip(0, 100)
```

```python
#Usage drop
df["Expected_Usage_7D"] = df["Usage_Minutes_30D"] / 30 * 7

df["Usage_Drop_Percent"] = (
    (df["Expected_Usage_7D"] - df["Usage_Minutes_7D"])
    / df["Expected_Usage_7D"]
) * 100

df["Usage_Drop_Percent"] = df["Usage_Drop_Percent"].clip(0, 100)
```

```python
#Support friction
df["Support_Friction_Index"] = (
    df["Support_Tickets"] * df["Avg_Resolution_Hours"]
)
```

```python
#Engagement score
df["Engagement_Score"] = (
    df["Login_Count_7D"] * 0.5
    + df["Usage_Minutes_7D"] * 0.3
    + df["NPS_Score"] * 0.2
)
```

```python
#Risk-related features
df["High_Inactivity"] = (df["Days_Since_Last_Login"] >= 14).astype(int)

df["Payment_Risk"] = (df["Payment_Delays"] >= 2).astype(int)

df["Low_Engagement"] = (df["Login_Drop_Percent"] >= 50).astype(int)
```

```python
new_features = [
    "Login_Drop_Percent",
    "Usage_Drop_Percent",
    "Support_Friction_Index",
    "Engagement_Score",
    "High_Inactivity",
    "Payment_Risk",
    "Low_Engagement"
]

df[new_features].describe()

#Save
df.to_csv("../Data/processed/customer_churn_features.csv", index=False)

print("Feature engineering completed!")
print("Shape:", df.shape)
```
